# Executive Summary

The authors of this study used computational text analysis to analyze the comments sections of ten red pill YouTube videos regarding mental health, therapy, and emotional vulnerability. The total number of comments from these ten YouTube videos was 13,075. Following the application of a filter for language relevant to mental health, approximately 1,000 of those comments were then analyzed using three different Natural Language Processing (NLP) methods (keyword frequency analysis, VADER sentiment analysis, and collocation analysis).

Four hypotheses were tested. Hypothesis 1 (H1) — that mentions of therapy are associated with a negative or dismissive sentiment — was supported, to a degree, by the average VADER score for comments about therapy being somewhat negative (≈ −0.04) and some of the major co-occurring words with "therapy" included the terms "weak," "do not" and "will not," although the overall distribution of sentiment for these comments varied from positive to negative. Hypothesis 2 (H2) — that the framing of male/male mental health through the lens of discipline and hierarchy — received the strongest level of support, as there were many co-occurring male-focused terms found within the context of "strong," "alpha," and "discipline," while the language used in describing vulnerability was not often used. Hypothesis 3 (H3) — that the framing of female/female mental health is invalidated or insufficiently framed — provided partial support for H3 based on the fact that the language used to describe females/female mental health had two different registers, specifically supportive (e.g., healing, support) and moralising (e.g., crazy, manipulative). Hypothesis 4 (H4) — that the language around self-improvement appears to exceed that of professional help-seeking behaviours — was not supported, since the comments made about therapy (n = 733) were made three times more often than self-improvement comments (n = 244).

The findings show that red pill communities don't just keep quiet when it comes to therapy; they also have a lot of arguments about therapy—specifically, how males and females can talk about their mental health and the role that societal standards of masculinity play in shaping discourse. This study shows the potential usefulness of using Natural Language Processing (NLP) methodologies to discover and explore hidden (or latent) ideological trends in Internet communications.

# 4. Data Cleaning and Preprocessing

```python
from googleapiclient.discovery import build
import os, pandas as pd

API_KEY = os.environ.get('YOUTUBE_API_KEY', "YOUR_KEY_HERE") 
youtube = build('youtube', 'v3', developerKey=API_KEY)

def get_comments(video_id, max_comments=1500):
    comments = []
    request = youtube.commentThreads().list(
        part='snippet', videoId=video_id,
        maxResults=100, textFormat='plainText'
    )
    while request and len(comments) < max_comments:
        response = request.execute()
        for item in response['items']:
            c = item['snippet']['topLevelComment']['snippet']
            comments.append({'video_id': video_id, 
                             'text': c['textDisplay'],
                             'likes': c['likeCount'], 
                             'date': c['publishedAt'],
                            "reply_count": item ["snippet"]["totalReplyCount"]
                            })
        request = youtube.commentThreads().list_next(request, response)
    return comments



# Proposed videos
video_ids = [
    "FaP_7hq6mmU",
    "6LDP_SyswNk",
    "79q5Pbaekds",
    "xO5AB1gCnRw",
    "VI6XWGKvUrE",
    "NX2ep5fCJZ8",
    "uMlRMAZ0lB0",
    "4nWcEtUV4do",
    "5yyWlp6sTv0",
    "dCCmdUNGukY"
]

all_comments = []

for vid in video_ids:
    print(f"Collecting from {vid}...")
    comments = get_comments(vid, max_comments=1500)
    print(f"Collected {len(comments)} comments")
    all_comments.extend(comments)

df = pd.DataFrame(all_comments)
df.to_csv("all_comments.csv", index=False)

print("Total comments collected:", len(all_comments))
df.shape
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)      
    text = re.sub(r"[^a-z\s]", "", text)     
    text = re.sub(r"\s+", " ", text).strip() 
    return text

df["clean_text"] = df["text"].apply(clean_text)
```

```python
import re

mental_health_keywords = [
    # Help-seeking
    "therapy", "therapist", "counseling", "help", "support",

    # Mental health conditions 
    "depression", "depressed", "anxiety", "anxious",
    "suicide", "suicidal", "mental health",

    # Emotional language 
    "lonely", "alone", "feelings", "emotion",

    # Masculinity / ideology 
    "masculinity", "alpha male", "gender roles",
    "self-improvement", "submissive woman",

    # Stigma / framing 
    "weak", "weak mind"
]

# Regex pattern
pattern = r"|".join(re.escape(keyword) for keyword in mental_health_keywords)

# Filtering
mh_df = df[df["clean_text"].str.contains(pattern, case=False, na=False)]

print("Mental health-related comments:", len(mh_df))
mh_df.head()
```

```python
def label_category(text):
    text = text.lower()
    categories = []

    # Therapy / help-seeking
    if any(word in text for word in ["therapy", "therapist", "counseling", "help", "support"]):
        categories.append("therapy")

    # Mental health conditions
    if any(word in text for word in ["depression", "depressed", "anxiety", "suicide", "suicidal"]):
        categories.append("mental_health_condition")

    # Masculinity / ideology
    if any(word in text for word in [
        "alpha", "alpha male", "masculinity",
        "gender roles", "self-improvement", "submissive"
    ]):
        categories.append("masculinity_discourse")

    # Emotional language
    if any(word in text for word in ["feelings", "emotion", "lonely", "alone"]):
        categories.append("emotional_expression")

    if not categories:
        categories.append("other")

    return categories


mh_df["category"] = mh_df["clean_text"].apply(label_category)

print("Mental health comments:", len(mh_df))
mh_df.head()
```

```python
import re

def count_mentions(texts, keywords):
    total_count = 0

    for text in texts:
        for word in keywords:
            matches = re.findall(rf"\b{word}\b", text)
            total_count += len(matches)

    return total_count


therapy_words = ['therapy', 'therapist', 'counseling', 'help', 'support']
self_improvement_words = ['gym', 'grind', 'discipline', 'money', 'success',
                          'alpha', 'strong', 'self improvement']

def count_mentions(texts, keywords):
    return sum(len(re.findall(rf'\b{w}\b', t)) for t in texts for w in keywords)

therapy_mentions = count_mentions(mh_df['clean_text'], therapy_words)
self_mentions    = count_mentions(mh_df['clean_text'], self_improvement_words)
print('Therapy-related mentions:', therapy_mentions)
print('Self-improvement mentions:', self_mentions)
```

# 6. Methodology

In this study 3 distinct but complementary computational linguistic approaches are used, they focus on different parts of how mental health conversations are had across 5 curriculum items. These methods range from broad quantitative descriptions of language to detailed contextual analyses of language use. 

**6.1 Frequency Analysis (Keywords)**

The vocabulary used to develop a baseline frequency in the corpus is therapy and self-improvement based and it was conducted using frequency analysis of keywords. This is done to perform the analysis using python through regular expressions (/b word boundaries). Count calculation using 2 levels of aggregating the data is done. With this we are able to see that the total occurences in each token showing how many times a word appears within the data, and the number of tokens which have at least 1 instance of any given keyword, with this method we are able to see a more conservative estimate of the themed prevelence. This method directly addresses the 4th Hypothesis. 

**6.2 VADER Sentiment Analysis**

In this study there is use of Valance Aware Dictionary and Sentiment Reasoner (VADER) for sentiment analyses of user-generated data on social media (YouTube). VADER is a lexicon-based method created by Hutto & Gilbert (2014) and was specifically designed to analyze text shared across social media platforms by creating an overall compound score between the additional positive/negative sentiment of each piece of text. The compound score ranges from -1 (extremely negative) to +1 (extremely positive). Based on the calculated compound scores sent via user-generated data, comments were classified as: positive (compound > 0.05), negative (compound < -0.05), and neutral (-0.05 ≤ compound ≤ 0.05) according to the thresholds proposed by VADER.

Each of the component analyzed by VADER incorporates positive, negative, and neutral valences within quantity and duration of use. Therefore, since VADER is designed to be an effective analysis tool that accurately captures informal and short-form internet text, it is effective at capturing these components of meaning for the sample set analyzed.

One of the bigger limitations with the use of tools like VADER is to analyse user-generated data in the context of data reviewed, due to its inability to understand concepts like sarcasm, irony, dismissive ideology accurately. For example; the phrase 'just go to the gym bro' has a very close to neutral VADER score even though that phrase's intention in a discursive manner is to dismiss the value of attaining therapy. This limitation is addressed by the inclusion of results in the collocation analysis which captures the context of the sample set that may not be effectively captures by the use of pure sentiment scoring.

```python
from collections import Counter

stopwords = set([
    "the","is","to","and","a","in","of","for","that","was","with",
    "as","on","it","this","be","are","at","by","an","or","from",
    "i","you","he","she","they","we","my","your","our"
])

def get_context_words(texts, target_word, window=3):
    words_list = []

    for text in texts:
        words = text.split()
        for i, word in enumerate(words):
            if word == target_word:
                start = max(0, i - window)
                end = i + window + 1
                context = words[start:end]

                for w in context:
                    if w not in stopwords and w != target_word:
                        words_list.append(w)

    return Counter(words_list)
```

# 8. Collocation and Context Analysis

With the use of collocation analysis, we are able to examine how gendered and therapy-related language is shown on the corpus by assessing linguistic context surrounding each time gendered terms or therapy-related terms appear. For these targets a sliding window of ±3 tokens is applied in the corpus. There is also a removal of NLTK stopwords and punctuation from the context window which results in a list of the common co-occuring terms that are sorted by frequency (i.e., words appearing within a window of ±3 tokens of each target). By doing this we are able to see an overview of the associated meanings given to these targets, which assists in identifying the themes in relation to the 2nd and 3rd Hypotheses. Collocation analysis in this example is particularly appropriate for identifying framing patterns that cannot be determined from sentiment alone. 

## Collocation Analysis By Gender

While Sentiment Analysis gives you an overall assessment of an Emotional Tone, it does not provide a full representation of how Masculinity and Femininity are being constructed Linguistically within the Discourse. To overcome this limitation, a Collocation Analysis was conducted to identify terms that would Collocate with Male and Female Related Terms within an analysis of the conversation about Mental Health, Vulnerability and Therapy and show how the Gender-Related Identities are framed.

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
```

```python
from collections import Counter
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words("english"))

def get_context_words(texts, target_terms, window_size=3):

    context_words = []

    for text in texts:

        words = nltk.word_tokenize(text)

        for i, word in enumerate(words):

            if word.lower() in target_terms:

                start = max(i - window_size, 0)
                end = min(i + window_size + 1, len(words))

                context = words[start:i] + words[i+1:end]

                cleaned_context = [

                    w.lower()

                    for w in context

                    if w.lower() not in stop_words
                    and w.lower() not in string.punctuation
                    and len(w) > 2
                ]

                context_words.extend(cleaned_context)

    return Counter(context_words)
```

# 10. Interpretation of Findings

Findings from the analysis indicate that there are substantial differences between females and males regarding how they communicate about therapy or mental health, and how they discuss gender, as indicated by frequency analysis, sentiment analysis, and collocation analysis.
In many instances, keywords used when referencing males were found to correlate strongly with terms such as emotional suppression, stigma and/or social pressure to appear strong, whereas, in comparison, many of the keywords used when referencing females were found to correlate more so with emotional inclusion (connection), support, and healing. 

These results are consistent with preexisting literature surrounding gender norms and the stigma surrounding mental health, especially concerning men's avoidance of vulnerability/emotional openness. These social norms may also assist in explaining why men's rate of therapy-seeking behaviour is lower than women. This analysis illustrates that Natural Language Processing (NLP) techniques are able to identify latent social patterns in the context of digital discussions about mental health and offer some insight into how gender can influence discourse regarding mental health.

VADER was developed based on general internet content; therefore, it does not perform well for text that includes sarcasm, irony, and “ideological dog whistling”, as is frequently the case in red pill-type content (e.g., the statement “just go to the gym bro” would be classified as neutral by VADER; however, functionally, that statement is being dismissive).

## 10.1 Therapy as contested discourse not silence

Perhaps the most theoretically relevant result from this analysis was that therapy is a frequent topic of discussion, and the H4s are often dismissed. Commenters who participate in these online spaces are actively and repeatedly engaging with the idea of therapy rather than either ignoring it or substituting other forms of self-help. This is consistent with research on online counter-discourse, which indicates that communities with strong ideological beliefs will produce considerable engagement around subjects with which they disagree (Ging, 2019). Because the number of instances of therapy identified in the data should not be perceived as endorsement, it may be more meaningful to appreciate these instances as an example of the ongoing ideological negotiation that continues to take place regarding the validity of therapy (e.g., getting support for oneself is often challenged, discussed or defended).

## 10.2 Masculine norms and emotional suppression

The collocation profile that characterizes male gender associations are mainly those words such as `strong', `alpha', `discipline' and stigma words (such as `weak'). These collocates illustrate how the masculine norms in the body of work at issue for men are very often comorbid with mental health (Addis & Mahalik 2003; Seidler et al. 2016) and also how through this discourse, emotional expression and seeking assistance indicate failures of self-sufficiency rather than rational acts in time of distress. Furthermore, it should be noted that this is through a relative contrast (i.e., `real' men are disciplined and strong while men who access therapy are `weak'). Connell (1995), states that this is an example of hegemonic masculinity, which is the dominant form of masculinity that subordinates emotional vulnerability.

## 10.3 Dual framing of female mental health

Research has suggested that female-referenced discourse includes both supportive and delegitimising language, which is consistent with what feminist discourse analysis suggest — women's emotional distress is often both pathologised and minimalized within a patriarchal discourse. Labels often used to refer to women's emotional distress, such as 'crazy', 'manipulative' and 'narcissistic', serve to embed a perception of women's mental health issues as character flaws or manipulative behaviour instead of representing a legitimate cause of distress. The use of moralising language often coexists with sympathetic language, which can indicate that the delegitimisation of women and their mental health in these contexts is not necessarily absolute but conditional: relatively sympathetic depiction of women exists in relation to how well women conform to normative relational roles.

## 10.4 Limitations of sentiment analysis in ideological contexts

The constrained support provided by H1 - as opposed to explicit support - demonstrates an inherent issue with the methodology. VADER makes decisions regarding sentiment polarity through surface level lexical elements alone, which means that it does not measure sarcasm, irony or the dismissive nature of apparent neutral statements. Thus, a request such as "men do not require therapy but only require exercise" will most likely be provided with a VADER score that is close to neutral or slightly positive due to the absence of vocabulary that is explicitly negative, while in reality it is an indirect statement rejecting the counsel of professionals. As such, lexicon based measurement systems for determining sentiment are not enough for the measurement of ideologically coded rhetoric and therefore collocation and manually annotated sentiment colour schemes are necessary additions.

# 11. Conclusion, Limitations and Recommendations

## 11.1 Conclusion

This study utilized natural language processing (NLP) techniques to analyze 1785 comments related to mental health across 10 YouTube videos that promote red pill ideology. It employed keyword frequency analysis, VADER sentiment analysis, and collocation analysis to examine the framing and discussion of therapy, emotional vulnerability, and mental health support by online communities.

The analysis also found that conversations within these online spaces are influenced by gender norms and shape how red pill ideology relates to mental health and therapy. Many men participating in these discussions expressed common themes associated with hegemonic notions of emotional repression or emotional suppression; emotional resilience or mental toughness; discipline or resilience toward overcoming emotional distress; and/or feeling pressure from society to conform to masculine norms related to emotions. In contrast, female users expressed themselves in a more fluid manner within a discursively contested space, where they expressed emotion; sought support from others; criticized and legitimized their emotional expression. The findings from this study suggest that the ways that people discuss emotions and/or emotions as they relate to seeking help reinforce gender-specific expectations of disclosure and help-seeking behavior online.

Therapy appears as an ongoing debate among these groups and indicates that it represents an ideologically-acceptable process of negotiation for them. Those commenting on therapy do so negatively, associated with weakness, dependence, and lack of self-control (e.g., characteristics not generally attributed to men). Results concerning issues related to therapy may exist beyond the analysis of computer-mediated discourse. The findings from this analysis are therefore likely more extensive than previously suggested; however, should the same findings be generalized to similarly defined red-pill-type online communities, they will likely continue to confirm negative stereotyping of men who seek therapy or suffer from emotional vulnerability. 

Thus, given the likelihood that men involved in red-pill-type communities are more likely than other men to find negative characteristics attributable to being therapeutic and/or emotionally vulnerable to be true, the implications of this study have repercussions for the manner in which men access services such as another type of therapy. This research demonstrates that a combination of natural-language processing (NLP) techniques (such as sentiment analysis, frequency analysis, and collocation analysis) have value in providing additional context for social relationships and ideologically based frames of reference in terms of masculine-oriented discourse about mental health in digital spaces.

## 11.2 Limitations

There are limitations to take into consideration when analyzing the results of this study.
VADER is not able to assess emotion or sentiment for many of the instances that it was designed to evaluate. It does not do a good job determining whether a post contains sarcasm, irony, or ideological dog-whistles (often found in “red pill” discourse). This could cause the sentiment scores to not be a reliable, stand-alone measure of dismissiveness.
The use of keywords to filter the corpus may lead to some selection bias. Comments that discuss mental health and do not contain the specified keywords would not be included in the study.
The corpus contained only 10 videos, which may not represent the entire diversity of red pill-type material. YouTube comment sections contain bots and artificially amplified activity; as such, the frequencies may be skewed.
The study examines language usage, not users — there is no way to know the demographics, genuineness, or mental health of individual commenters from the data alone.

## 11.3 Recommendations

There are many ways that future studies could address each of the limitations in this article. Latent Dirichlet Allocation (LDA) topic modelling would allow for unsupervised identification of thematic clusters beyond the keyword categories that were defined in this study and may reveal new patterns of discourse that were not predicted in the initial hypothesis. Manual coding by trained reviewers on a sample of random comments would validate the findings regarding sentiment and collocations and help produce an industry-specific sentiment lexicon that is better aligned with red pill discourse. Longitudinal data (i.e., looking at the same channels over several years) could be examined to see if there have been changes in therapy discourse across these platforms as a result of increased public discourse surrounding men's mental health. Finally, a cross-platform comparison (i.e., comparing comments from Reddit and TikTok) would test whether these patterns generalise to other platforms in addition to YouTube.
